```python
dbutils.widgets.text("catalog", "workspace")
dbutils.widgets.text("schema", "pipeline_breweries")

catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")

current_page = dbutils.jobs.taskValues.get(
    taskKey = "Compute_Page",
    key = "current_page",
    default = 1
)

config_table = f"{catalog}.{schema}.bronze_breweries_job_config"

#Salvare in una Delta table di configurazione
spark.createDataFrame( [ ( int(current_page), ) ] , ["current_page"]) \
     .write.mode("overwrite") \
     .saveAsTable(config_table)
```
